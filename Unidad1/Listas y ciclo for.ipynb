```python
#### listas
#las listas siepre empenzan desde el 0
Alum=["ana","andres","carlos"]
print(Alum)
print(Alum[0])
print(Alum[1])
print(Alum[2])

print("-----------------")

#ciclo for para mostra valores de la lista

for i  in  range(3):

  print(Alum[i])

print("-----------------")

#lista negativos
#se puede acceder a los valores con un valor negativo
Alum=["ana","andres","carlos"]

print(Alum[-2])









```

```python


#podemos modificar el valor de u elemento en la lista asignando un nuvo valor al indice
print(Alum)

Alum[1]="jean"
print(Alum)
```

```python
#metodo append para añadir elemento a la lista

print(Alum)
Alum.append("juanitoXD")
print(Alum)


#metodo insert para añadir elemeto en posicion especifica

Alum.insert(2,"josesito")
print(Alum)

```

```python
print(Alum)
#medodo remove borra un elemento de la lista especifiacando el valor, si existen valores repetidos solo borrara el primer valor
Alum.remove("juanitoXD")

print(Alum)
```

```python

#metodo "pop" borra elemento en posicion especifica
print(Alum)
Alum.pop(2)
print(Alum)
```

```python

#metodo "pop" borra elemento en posicion especifica
print(Alum)
del Alum[2]
print(Alum)
```

```python
#la fuuncion len() me permite conocer el numero de items en la lisat
print("numero de elementos en la lista es de",len(Alum))



```
